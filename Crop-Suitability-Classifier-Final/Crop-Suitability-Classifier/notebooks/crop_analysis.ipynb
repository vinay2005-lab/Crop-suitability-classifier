# CropSuit — Crop Suitability Classification

This notebook documents the capstone workflow.

```python
from pathlib import Path
import sys
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path.cwd()
if not (ROOT / "data").exists() and (ROOT.parent / "data").exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT))

from src.download_dataset import download_dataset
from src.preprocessing import load_and_clean_data, FEATURES, TARGET

data_path = ROOT / "data" / "Crop_recommendation.csv"
if not data_path.exists():
    download_dataset()

df, duplicates_removed = load_and_clean_data(data_path)
print("Shape:", df.shape)
print("Duplicates removed:", duplicates_removed)
df.head()
```

## Data quality inspection

```python
print(df.info())
print("\nMissing values:")
print(df.isna().sum())
print("\nClass counts:")
print(df[TARGET].value_counts())
print("\nSummary statistics:")
display(df[FEATURES].describe())
```

## Exploratory Data Analysis

```python
plt.figure(figsize=(12, 7))
df[TARGET].value_counts().sort_values().plot(kind="barh")
plt.title("Crop Class Distribution")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 7))
plt.imshow(df[FEATURES].corr(), cmap="coolwarm", aspect="auto"); plt.colorbar()
plt.title("Feature Correlation Matrix")
plt.tight_layout()
plt.show()
```
